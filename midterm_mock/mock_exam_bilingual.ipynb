````python
def read_data(filename):
    # TODO: ファイルを読み込み、文字列として返す
    # TODO: Read the file and return its contents as a string.
    with open(filename, encoding="utf-8") as file:
        text = file.read()
        return text

    
````

````python
text = read_data("login_data.csv")
print(text)
````

## 課題2：利用する行を取り出す / Task 2: Select the data lines

文字列を行ごとに分け、ヘッダーと空行を除いたリストを返してください。

Split the text into lines and return a list without the header or blank lines.

````python
print(text)
print(text.splitlines())
lines=[]
for i,line in enumerate(text.splitlines()):
    if i==0:
        continue
    elif line=="":
        continue
    else:
        lines.append(line)

print(lines)
````

````python
def get_data_lines(text):
    # TODO: ヘッダーと空行を除いた行のリストを返す
    # TODO: Return a list of lines without the header or blank lines.

    return [line for line in text.splitlines()[1:] if line != ""]
````

````python
lines = get_data_lines(text)
print(lines)
````

## 課題3：アクティブなユーザー名を取り出す / Task 3: Extract active usernames

各行をカンマで分割し、`status`が`active`の行からユーザー名を取り出してください。

Split each line by commas. Extract the username from each row whose `status` is `active`.

````python
def get_active_usernames(lines):
    # TODO: activeの行からユーザー名を取り出し、リストで返す
    # TODO: Extract usernames from active rows and return them as a list.
    activeusers=[]
    for user in lines:
        if user.split(",")[3]=="active":
            activeusers.append(user.split(",")[0])
    return activeusers
````

````python
names = get_active_usernames(lines)
print(names)
````

## 課題4：並べ替えて結果を表示する / Task 4: Sort and display the results

ユーザー名を昇順に表示し、セッション数と重複しないユーザー数を表示してください。

Display the usernames in ascending order. Then display the number of sessions and the number of unique users.

````python
def show_result(names):
    # TODO: 指定された形式で結果を表示する
    # TODO: Display the results in the specified format.
    sorted_names=sorted(names)

    for v in sorted_names:
        print(v)
    
    print(len(names))
    print(len(set(names)))
````

````python
show_result(names)
````
